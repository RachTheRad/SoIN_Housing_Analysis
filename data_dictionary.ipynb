## Data Dictionary
### Southern Indiana Housing and Public Education Analysis


County Market Tracker - (Source - Redfin Data Center)[https://www.redfin.com/news/data-center/]
|Column Name | Data Type | Description |
|-------------|-----------|---------------------------------------|
|period_begin |	Object | Beginning date data in this row was pulled from | 
|period_end	| Object | End date data in this row was pulled from |
|table_id | Int64 | Table ID key |
|region	| Object | Counties within the states from the USA |
|state | Object | States from the USA |
|state_code | Object | Numeric ID assigned to each state by Redfin |
|property_type| Object | What kind of property was being sold (ex: Condo/Co-op, Townhouse, Single Family Residential, etc.) |
|property_type_id | Int64 |	Numeric code given to the property type (assuming by Redfin) |
|median_sale_price | Float64 | The median sale price for this county for the property type listed on the same row for the region listed. |
|median_list_price | Float64 |The median list price for this county for the property type listed on the same row for the region listed.
|median_ppsf | Float64 | Median price per square foot for property sold. |
|median_ppsf_yoy | Float64 | Median price per square foot for property sold for year over year. |
|median_list_ppsf |	Float64 |Median list price per square foot for property sold. |
|median_list_ppsf_yoy |	Float64 | Median lsit price per square foot for property sold year over year. |
|inventory | Float64 | Number of real estate listings available. |
|inventory_yoy | Float64 | Inventory change year over year, listed as a percentage point. |
|sold_above_list | Float64 | Properties that sold above list price. |
|parent_metro_region | Objcet | Closest metro region. If there is no close metro region it is stated as "[State] nonmetropolitan area." |
|parent_metro_region_metro_code | Float64 | Numeric ID assigned to the parent metro region. |
|last_updated | Object | Date the data was last updated. |
