```python
"""
Cleans and processes the Bartholomew, Clark, and Floyd County Housing DataFrame.

    This function performs the following operations on the input DataFrame:
    1. Cleans column names (capitalizes and removes underscores).

    Parameters:
        bart_clark_floyd (pandas.DataFrame): The input housing dataset.

    Returns:
        pandas.DataFrame: The cleaned and processed DataFrame.

    Example:
        >>> clean_kick_starter = clean_ks(bart_clark_floyd)\"
"""
            
def clean_hd (bart_clark_floyd) -> pd.DataFrame:

    # Clean column names
    bart_clark_floyd.columns =  bart_clark_floyd.columns.str.title()
    bart_clark_floyd.columns =  bart_clark_floyd.columns.str.strip().str.replace('_', ' ')


    return bart_clark_floyd

bart_clark_floyd_cleaned = clean_hd(bart_clark_floyd)

display(bart_clark_floyd_cleaned)
```

```python
def summarize_housing_data(bart_clark_floyd_cleaned: DataFrame) -> str:
    """
    Calculates and summarizes the total, mean, and median number of backers 
    from a Housing dataset.

    Parameters:
        bart_clark_floyd_cleaned (pandas.DataFrame): The housing dataset containing a 
                                         'Median Sale Price' column.

    Returns:
        str: A formatted string summarizing the total, mean (rounded to 2 decimal places),
             and median number of backers.
    """
    # Calculate total, mean, and median backers
    sales_min = bart_clark_floyd_cleaned['Median Sale Price'].min()
    sales_mean = bart_clark_floyd_cleaned['Median Sale Price'].mean().round(2)
    sales_median = bart_clark_floyd_cleaned['Median Sale Price'].median()
    sales_max = bart_clark_floyd_cleaned['Median Sale Price'].max()

    # Return a formatted string
    return (f"The mean sales price is {sales_mean}.\n"
            f"The median sales price is {sales_median}.\n"
            f"The min sales price is {sales_min}.\n"
            f"The max sales price is {sales_max}")

sales_summed = summarize_housing_data(bart_clark_floyd_cleaned)
print(sales_summed)
```

```python
def summarize_ppsf(bart_clark_floyd_cleaned: DataFrame) -> str:
    """
    Calculates and summarizes the mean and median pricing per squarefoot 
    from a Housing dataset.

    Parameters:
        bart_clark_floyd_cleaned (pandas.DataFrame): The housing dataset containing a 
                                         'Median Ppsf' column.

    Returns:
        str: A formatted string summarizing the mean (rounded to 2 decimal places)
             and median price per square foot of listed houses. 
    """
    # Calculate total, mean, and median backers
    ppsf_mean = bart_clark_floyd_cleaned['Median Ppsf'].mean().round(2)
    ppsf_median = bart_clark_floyd_cleaned['Median Ppsf'].median()

    # Return a formatted string
    return (f"The mean price per square foot is {ppsf_mean}.\n"
            f"The median price per square foot is {ppsf_median}.")

ppsf_summed = summarize_ppsf(bart_clark_floyd_cleaned)
print(ppsf_summed)
```
